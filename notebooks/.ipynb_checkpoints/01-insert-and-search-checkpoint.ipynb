# pgvector I: storing and searching embeddings

Postgres as a semantic-search engine. In this notebook we:

1. create a table with **normal columns plus one `vector` column**,
2. embed a small bilingual corpus with **`multilingual-e5-small`** (384 dims, CPU-friendly),
3. **insert** the embeddings and **retrieve** them with exact nearest-neighbor SQL,
4. use **metadata columns to scope** a search.

Stack: Postgres 16 + pgvector 0.8.x, `psycopg` 3, `sentence-transformers`. This is the mechanics behind U2T03, where the same operations run behind a queue and a UI.

```python
import os, warnings
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import psycopg
from pgvector.psycopg import register_vector

conn = psycopg.connect(os.environ["DATABASE_URL"], autocommit=True)
register_vector(conn)   # teaches this connection to bind/fetch the vector type
cur = conn.cursor()

cur.execute("SELECT extversion FROM pg_extension WHERE extname = 'vector'")
print("pgvector", cur.fetchone()[0])
```

## 1. The table: ordinary columns + one vector column

A vector column is just another type, but its **dimension is fixed at DDL time** (storage layout depends on it), so one column cannot host two models with different dims. That is why a multi-model system (U2T03's reference impl) uses one embedding table per model.

```python
cur.execute("DROP TABLE IF EXISTS documents")
cur.execute("""
    CREATE TABLE documents (
        id        SERIAL PRIMARY KEY,
        title     TEXT NOT NULL,
        category  TEXT NOT NULL,        -- metadata used to scope searches
        lang      TEXT NOT NULL,        -- 'en' | 'es'
        year      INT  NOT NULL,
        content   TEXT NOT NULL,
        embedding vector(384) NOT NULL  -- e5-small output
    )
""")

cur.execute("""
    SELECT ordinal_position, column_name, udt_name, is_nullable
    FROM information_schema.columns
    WHERE table_name = 'documents' ORDER BY ordinal_position
""")
pd.DataFrame(cur.fetchall(), columns=[d.name for d in cur.description])
```

## 3. Embed with e5-small

E5 is asymmetric: stored documents get the `passage:` prefix, queries get `query:`. Vectors are L2-normalized, so cosine similarity is a dot product and all three pgvector distance operators rank identically (we check that below). First run downloads the model (~470 MB); after that it is cached offline.

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("intfloat/multilingual-e5-small")
texts = ["passage: " + row[4] for row in corpus]
emb = model.encode(texts, normalize_embeddings=True, show_progress_bar=False)

print(type(emb).__name__, emb.shape, "norms ~", np.linalg.norm(emb, axis=1).round(6).min(),
      "to", np.linalg.norm(emb, axis=1).round(6).max())
```

## 4. Insert

`register_vector` lets us bind NumPy arrays directly. What actually travels over the wire is the literal text form `'[0.013, -0.117, ...]'`, the same thing you would type in `psql`. We insert with `executemany` (36 rows is nothing; notebook 02 uses `COPY` for 50k).

```python
cur.execute("DELETE FROM documents")
cur.executemany(
    "INSERT INTO documents (title, category, lang, year, content, embedding)"
    " VALUES (%s, %s, %s, %s, %s, %s)",
    [(t, c, l, y, s, e) for (c, l, y, t, s), e in zip(corpus, emb)],
)
cur.execute("SELECT count(*) FROM documents")
print("rows:", cur.fetchone()[0])

# the wire format, and what a stored row looks like
cur.execute("SELECT '[0.1, 0.2, 0.3]'::vector")
print("literal ->", cur.fetchone()[0])
cur.execute("SELECT left(embedding::text, 42) || '...' FROM documents WHERE id = 1")
print("stored  ->", cur.fetchone()[0])
```

## 5. Exact nearest-neighbor search

`<=>` is cosine **distance** (1 - cosine similarity): smaller is closer. Without any index this is an exact sequential scan, and at 36 rows that is exactly what you want.

```python
def search(query, k=5, where="", params=()):
    qv = model.encode(["query: " + query], normalize_embeddings=True)[0]
    sql = (f"SELECT id, title, category, lang, 1 - (embedding <=> %s) AS similarity"
           f" FROM documents {where} ORDER BY embedding <=> %s LIMIT {k}")
    cur.execute(sql, (qv, *params, qv))
    return pd.DataFrame(cur.fetchall(), columns=[d.name for d in cur.description])

search("how do I speed up a slow query")
```

Index tuning, EXPLAIN and VACUUM rows surface, as they should. Note the score is a cosine similarity to a 384-d point, not a percentage of anything.

## 6. The same query in Spanish

Nothing changes on the database side. The multilingual encoder put both languages in one space, so `cómo acelerar consultas lentas` retrieves the same row, and its Spanish twin.

```python
search("cómo acelerar consultas lentas")
```

The Spanish query puts the right row on top with a clear margin, without a single shared keyword. Look at the tail too: it drifts toward other Spanish rows even when off-topic. Multilingual spaces mix "same meaning" with "same language" a little; rank survives it, but it is a real effect worth knowing when you read these tables.

## 7. Three operators, one ranking

| operator | distance | index opclass |
|---|---|---|
| `<=>` | cosine | `vector_cosine_ops` |
| `<->` | euclidean (L2) | `vector_l2_ops` |
| `<#>` | negative inner product | `vector_ip_ops` |

Because these embeddings are unit vectors, all three order the table identically. The operator choice matters for the **index** (notebook 02): an HNSW index is built for one operator class.

```python
qv = model.encode(["query: how do I speed up a slow query"], normalize_embeddings=True)[0]
frames = {}
for name, op in [("cosine <=>", "<=>"), ("euclidean <->", "<->"), ("inner product <#>", "<#>")]:
    cur.execute(f"SELECT id FROM documents ORDER BY embedding {op} %s LIMIT 5", (qv,))
    frames[name] = [r[0] for r in cur.fetchall()]
pd.DataFrame(frames)
```

## 8. Metadata filtering: shrinking the scope

Plain SQL `WHERE` composes with vector search. The query below is the same "slow query" question, scoped to `category = 'embeddings'`: now the speed story is Matryoshka truncation and ANN, not B-trees. Exact search plus a filter is always correct; the subtle failure mode only appears once an **approximate** index meets a filter (notebook 02, section 5). This `WHERE active` + top-k pattern is precisely U2T03's search interface.

```python
display(search("how do I speed up a slow query"))
display(search("how do I speed up a slow query",
               where="WHERE category = %s", params=("embeddings",)))
```

## 9. Reading the scores honestly

Three calibration facts for any similarity table you will ever show:

1. identical text scores exactly 1.0 against itself (same prefix),
2. `query: text` vs `passage: text` for the same text scores *less* than 1.0: the prefixes are different points, which is the asymmetry made visible,
3. random document pairs already sit near 0.8 on this corpus (measured 0.81). e5-small's scores are high and narrow, so **rank is the signal, not the absolute number**.

```python
qv = model.encode(["query: how do I speed up a slow query"], normalize_embeddings=True)[0]
pv = model.encode(["passage: how do I speed up a slow query"], normalize_embeddings=True)[0]
qq = model.encode(["query: how do I speed up a slow query"], normalize_embeddings=True)[0]
print(f"query vs query   (same prefix, same text): {qv @ qq:.4f}")
print(f"query vs passage (same text, both sides):  {qv @ pv:.4f}")

r = np.random.default_rng(0)
pair = [float(emb[i] @ emb[j]) for i, j in (r.choice(len(emb), 2, replace=False) for _ in range(200))]
print(f"random doc-doc pairs: mean {np.mean(pair):.3f}, min {min(pair):.3f}, max {max(pair):.3f}")
```

**Wrong-configuration footnote.** We also ran the prefix-dropped configuration on this corpus: the top-score penalty stayed within ±0.006 across ten queries, matching what the U2T03 reference implementation measured on e5-base (0.9011 vs 0.9007 on its probe query). The penalty is configuration- and query-dependent, which is exactly why U2T03 asks you to *measure* a wrong run instead of assuming it fails.
